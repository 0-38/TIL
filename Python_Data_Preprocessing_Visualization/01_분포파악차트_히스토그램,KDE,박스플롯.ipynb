## 1. 히스토그램

### 1-1 ) 히스토그램 구조
#### "선수들의 주간 훈련시간 분포를 히스토그램으로 시각화"

```python
import matplotlib.pyplot as plt
%config InlineBackend.figure_format = 'retina'

training_hours = [
    40, 42, 44, 46, 48,
    50, 52, 54, 56, 58
]

plt.hist(
    training_hours,
    bins=[40, 45, 50, 55, 60],
    edgecolor="white"
)

plt.xlabel("Training Hours")
plt.ylabel("Frequency")
plt.title("Weekly Training Hours Distribution")
```

### 1-4 ) bins 값에 따른 차이 비교
#### "같은 보행속도 데이터를 bins=5와 bins=10으로 비교"

```python
import matplotlib.pyplot as plt

gait_speed = [
    0.9, 1.0, 1.0, 1.1, 1.1,
    1.2, 1.2, 1.2, 1.3, 1.3,
    1.4, 1.4, 1.5, 1.5, 1.6
]

fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes = axes.flatten()

for ax, bins in zip(axes, range(5, 25, 5)):

    ax.hist(gait_speed, bins=bins, edgecolor="white")

    ax.set_title(f"Gait Speed - bins={bins}")
    ax.set_xlabel("Speed")
    ax.set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```

### 1-6 ) Matplotlib 한글 폰트 설정 

```python
import matplotlib.pyplot as plt
import platform

# 한글폰트 설정
if platform.system() == "Windows":
    plt.rcParams["font.family"] = "Malgun Gothic"

elif platform.system() == "Darwin":
    plt.rcParams["font.family"] = "AppleGothic"

plt.rcParams["axes.unicode_minus"] = False

# 데이터프레임 생성
gait_speed = [
    0.9, 1.0, 1.0, 1.1, 1.1,
    1.2, 1.2, 1.2, 1.3, 1.3,
    1.4, 1.4, 1.5, 1.5, 1.6
]

plt.hist(
    gait_speed,
    bins=6,
    edgecolor="white"
)

plt.xlabel("보행 속도")
plt.ylabel("빈도")
plt.title("보행속도 분포도")
```

## 2. KDE(Kernel Density Estimation)

### 2-2 ) 두 그룹의 분포 비교
#### "두 육상팀의 훈련시간 분포를 KDE로 비교"

```python
import seaborn as sns
import matplotlib.pyplot as plt

team_a = [
    40, 42, 44, 45, 46,
    48, 50, 52, 53, 55,
    56, 58, 60, 62, 64
]

team_b = [
    50, 52, 54, 55, 57,
    58, 60, 62, 63, 65,
    67, 68, 70, 72, 75
]

sns.kdeplot(
    team_a,
    label="Team A",
    fill=True
)

sns.kdeplot(
    team_b,
    label="Team B",
    fill=True,
    color="green"
)

plt.xlabel("Training Time")
plt.ylabel("Density")
plt.title("Training Time Distribution by Team")
plt.legend()

plt.show()
```

## 3. 박스플롯(Box Plot)

### 3-2 ) 박스플롯으로 이상치 후보 확인
#### "운동 수행시간의 박스플롯 생성"

```python
import matplotlib.pyplot as plt

work_times = [
    20, 21, 22, 23, 23,
    24, 24, 25, 25, 26,
    26, 27, 27, 28, 29,
    30, 31, 32, 65, 72
]

plt.boxplot(
    work_times
)

plt.ylabel("Time")
plt.title("Performance Time Box Plot")

plt.show()
```

## 4. 히스토그램 · KDE · 박스플롯 비교

### 4-1 ) 동일한 데이터를 세 차트로 비교

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 데이터 생성
delivery_times = [
    1.2, 1.5, 1.7, 1.8, 2.0,
    2.0, 2.1, 2.2, 2.3, 2.4,
    2.5, 2.6, 2.7, 2.8, 3.0,
    3.1, 3.2, 3.5, 6.5, 8.0
]

# subplot 생성
fig, axes = plt.subplots(
    nrows=1,
    ncols=3,
    figsize=(12, 4)
)

# 1. Histogram
axes[0].hist(
    delivery_times,
    bins=5,
    edgecolor="white"
)
axes[0].set_title("Histogram")

# 2. KDE
sns.kdeplot(
    delivery_times,
    fill=True,
    ax=axes[1]
)
axes[1].set_title("KDE")

# 3. Box Plot
axes[2].boxplot(
    delivery_times
)
axes[2].set_title("Box Plot")

# 그래프 간격 자동 조절
plt.tight_layout()

plt.show()
```
